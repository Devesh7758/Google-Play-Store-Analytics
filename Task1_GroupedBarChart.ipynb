```python
apps_df = pd.read_csv(r"C:\Users\ashis\OneDrive\Desktop\GooglePlayProject\Play Store Data.csv")
```

```python
apps_df.head()
```

```python
apps_df['Size'].unique()
```

```python
apps_df = apps_df[apps_df['Size'].str.contains('M', na=False)]
```

```python
apps_df['Size'] = apps_df['Size'].str.replace('M', '')
```

```python
apps_df['Size'] = apps_df['Size'].astype(float)
```

```python
apps_df = apps_df[apps_df['Size'] > 10]
```

```python
apps_df[['App','Size']].head()
```

```python
apps_df = apps_df[apps_df['Rating'] > 4.0]
```

```python
apps_df[['App','Rating']].head()
```

```python
apps_df['Last Updated'] = pd.to_datetime(apps_df['Last Updated'])
```

```python
apps_df = apps_df[apps_df['Last Updated'].dt.month == 1]
```

```python
apps_df[['App','Last Updated']].head()
```

```python
apps_df['Installs'].head()
```

```python
apps_df['Installs'] = apps_df['Installs'].str.replace(',', '')
```

```python
apps_df['Installs'] = apps_df['Installs'].str.replace('+', '')
```

```python
apps_df['Installs'] = apps_df['Installs'].astype(int)
```

```python
apps_df[['App','Installs']].head()
```

```python
grouped_df = apps_df.groupby('Category').agg({
    'Installs':'sum',
    'Rating':'mean',
    'Reviews':'sum'
}).reset_index()
```

```python
grouped_df.head()
```

```python
top_10_categories = grouped_df.sort_values(by='Installs',ascending=False).head(10)
```

```python
top_10_categories
```

```python
apps_df['Reviews'] = pd.to_numeric(apps_df['Reviews'], errors='coerce')
```

```python
grouped_df = apps_df.groupby('Category').agg({
    'Installs': 'sum',
    'Rating': 'mean',
    'Reviews': 'sum'
}).reset_index()
```

```python
grouped_df = grouped_df[grouped_df['Rating'] >= 4.0]
```

```python
top_10_categories = grouped_df.sort_values(by='Installs',ascending=False).head(10)
```

```python
top_10_categories
```

```python
fig1 = go.Figure()
fig1.add_trace(go.Bar(
    x=top_10_categories['Category'],
    y=top_10_categories['Rating'],
    name='Average Rating'
))
fig1.add_trace(go.Bar(
    x=top_10_categories['Category'],
    y=top_10_categories['Reviews'],
    name='Total Reviews'
))
fig1.update_layout(
    title='Average Rating and Reviews by Category',
    xaxis_title='Category',
    yaxis_title='Values',
    barmode='group',
    template='plotly_dark',
    width=1200,
    height=600
)
fig1.show()
```

```python
from datetime import datetime
import pytz
ist = pytz.timezone('Asia/Kolkata')
current_time = datetime.now(ist)
current_hour = current_time.hour
show_graph = current_hour >= 15 and current_hour < 17
print("Current IST Time:", current_time.strftime("%H:%M:%S"))
print("Show Graph:", show_graph)
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go
fig1 = make_subplots(specs=[[{"secondary_y": True}]])
fig1.add_trace(
    go.Bar(
        x=top_10_categories['Category'],
        y=top_10_categories['Reviews'],
        name='Total Reviews'
    ),
    secondary_y=False
)
fig1.add_trace(
    go.Scatter(
        x=top_10_categories['Category'],
        y=top_10_categories['Rating'],
        name='Average Rating',
        mode='lines+markers'
    ),
    secondary_y=True
)
fig1.update_layout(
    title='Average Rating and Reviews by Category',
    template='plotly_dark',
    width=1100,
    height=600,
    xaxis_title='Category',
    xaxis_tickangle=-25
)
fig1.update_yaxes(
    title_text='Total Reviews',
    secondary_y=False
)
fig1.update_yaxes(
    title_text='Average Rating',
    secondary_y=True
)
fig1.update_yaxes(
    showgrid=False,
    secondary_y=True
)
fig1.show()
fig1.write_html("task1.html")
```
